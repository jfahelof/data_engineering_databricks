### Agora, iremos visualizar em gráficos para facilitar a 

```python
import plotly.express as px
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
fig = px.bar(
    df_bairro_pandas.sort_values(by="total_acidentes"),
    x="total_acidentes",
    y="bairro",
    orientation="h",
    text="total_acidentes",
    hover_data={
        "total_vitimas": True,
        "total_fatais": True,
        "taxa_fatalidade": ':.2f'
    },
    title="Acidentes por Bairro"
)

fig.update_layout(
    xaxis_title="Número de Acidentes",
    yaxis_title="Bairro",
    height=800 
)

fig.show()
```
